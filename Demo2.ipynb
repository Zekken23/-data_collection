```python
pengiriman = [
    {"resi": "A001", "berat": 2.5, "ongkir_perkg": 10000, "status": "Terkirim"},
    {"resi": "A002", "berat": 5, "ongkir_perkg": 12000, "status": "Dalam Proses"},
    {"resi": "A003", "berat": 1.2, "ongkir_perkg": 10000, "status": "Terkirim"},
    {"resi": "A004", "berat": 10, "ongkir_perkg": 9000, "status": "Terkirim"},
    {"resi": "A005", "berat": 0.5, "ongkir_perkg": 15000, "status": "Batal"},
    {"resi": "A006", "berat": 3.8, "ongkir_perkg": 11000, "status": "Dalam Proses"},
    {"resi": "A007", "berat": 7.2, "ongkir_perkg": 9500, "status": "Terkirim"},
    {"resi": "A008", "berat": 4.5, "ongkir_perkg": 10500, "status": "Terkirim"},
    {"resi": "A009", "berat": 6.0, "ongkir_perkg": 9800, "status": "Batal"},
    {"resi": "A010", "berat": 9.1, "ongkir_perkg": 8700, "status": "Terkirim"},
    {"resi": "A011", "berat": 2.0, "ongkir_perkg": 12500, "status": "Dalam Proses"},
    {"resi": "A012", "berat": 1.5, "ongkir_perkg": 14000, "status": "Terkirim"},
    {"resi": "A013", "berat": 8.3, "ongkir_perkg": 9200, "status": "Terkirim"},
    {"resi": "A014", "berat": 0.8, "ongkir_perkg": 16000, "status": "Batal"},
    {"resi": "A015", "berat": 5.6, "ongkir_perkg": 9900, "status": "Terkirim"}
]
```

```python
# Muhammad Yusron Al Ghoni Rizqullah
# 202310370311333

from functools import reduce

# 1. Filter data pengiriman yang statusnya terkirim saja
def is_terkirim(item):
    return item["status"] == "Terkirim"

pengiriman_terkirim = filter(is_terkirim, pengiriman)

print("Skenario 1: Pengiriman Terkirim")
for item in pengiriman_terkirim:
    print(item)
```

```python
# Muhammad Yusron Al Ghoni Rizqullah
# 202310370311333

# 2. Tambahkan informasi total ongkir pada data yang terkirim tadi (menggunakan map)
# We need to re-filter as the filter object is consumed after iteration
pengiriman_terkirim_for_map = filter(is_terkirim, pengiriman)

def calculate_total_ongkir(item):
    item["total_ongkir"] = item["berat"] * item["ongkir_perkg"]
    return item

pengiriman_dengan_ongkir = map(calculate_total_ongkir, pengiriman_terkirim_for_map)

print("\nSkenario 2: Pengiriman Terkirim dengan Total Ongkir")
for item in pengiriman_dengan_ongkir:
    print(item)
```

```python
# Muhammad Yusron Al Ghoni Rizqullah
# 202310370311333

# We need to re-map as the map object is consumed after iteration
pengiriman_terkirim_for_reduce = filter(is_terkirim, pengiriman)
pengiriman_dengan_ongkir_for_reduce = map(calculate_total_ongkir, pengiriman_terkirim_for_reduce)


def sum_total_ongkir(total, item):
    return total + item["total_ongkir"]

total_pendapatan = reduce(sum_total_ongkir, pengiriman_dengan_ongkir_for_reduce, 0)

print("\nSkenario 3: Total Pendapatan dari Pengiriman Terkirim")
print("Total Pendapatan:", total_pendapatan)
```

```python
# Muhammad Yusron Al Ghoni Rizqullah
# 202310370311333

# 4. Siapkan data semua pengiriman yang tidak dibatalkan menggunakan list comprehension
# dan siapkan fungsi rekursif untuk menghitung total berat nya

def is_not_batal(item):
    return item["status"] != "Batal"

pengiriman_tidak_dibatalkan = [item for item in pengiriman if is_not_batal(item)]

print("\nSkenario 4: Pengiriman Tidak Dibatalkan (List Comprehension)")
for item in pengiriman_tidak_dibatalkan:
    print(item)

# Fungsi rekursif untuk menghitung total berat
def hitung_total_berat_rekursif(data_list):
    if not data_list: # Base case: empty list
        return 0
    else: # Recursive case: add weight of first item and call function on rest of the list
        return data_list[0]["berat"] + hitung_total_berat_rekursif(data_list[1:])

total_berat_tidak_dibatalkan = hitung_total_berat_rekursif(pengiriman_tidak_dibatalkan)

print("\nTotal berat pengiriman tidak dibatalkan (Rekursif):", total_berat_tidak_dibatalkan)
```
